```python

class plan_to_d3(dspy.Signature):
    """ """
    plan = dspy.InputField(desc="Planner instructions around what they want to visualize")
    dataset_context = dspy.InputField(desc="Information about the dataset, columns, stats")
    styling_instructions = dspy.InputField(desc="Styling instructions for the charts / dashboard")
    d3_js_code = dspy.OutputField(desc="d3.js code that the js frontend can run to visualize")


# query_enhancer = dspy.Predict("original_query, dataset_context ->chart_categories:List[Literal['line_charts','bar_charts','histograms','pie_charts','tabular_and_generic_charts','no_chart']]")

plan_to_d3_sign = dspy.Predict(plan_to_d3)

styling_instructions = [
    {
        "category": "line_charts",
        "description": "Used to visualize trends and changes over time, often with multiple series.",
        "styling": {
            "theme": "light",  # replaces 'template'
            "axes": {
                "stroke_width": 0.2,
                "grid_stroke_width": 1,
                "grid_color": "#e0e0e0"
            },
            "title": {
                "bold_html": True,
                "include": True
            },
            "colors": "use color scale (d3.schemeCategory10 or interpolated if >10 lines)",
            "annotations": ["min", "max"],
            "number_format": {
                "apply_k_m": True,
                "thresholds": {"K": 1000, "M": 100000},
                "percentage_decimals": 2,
                "percentage_sign": True
            },
            "svg_defaults": {"height": 600, "width": 800, "margin": {"top": 50, "right": 50, "bottom": 50, "left": 60}}
        }
    },
    {
        "category": "bar_charts",
        "description": "Useful for comparing discrete categories or groups with bars representing values.",
        "styling": {
            "theme": "light",
            "axes": {
                "stroke_width": 0.2,
                "grid_stroke_width": 1
            },
            "title": {"bold_html": True, "include": True},
            "annotations": ["bar values"],
            "bar_style": {"padding": 0.1, "corner_radius": 3},
            "number_format": {
                "apply_k_m": True,
                "thresholds": {"K": 1000, "M": 100000},
                "percentage_decimals": 2,
                "percentage_sign": True
            },
            "svg_defaults": {"height": 600, "width": 800}
        }
    },
    {
        "category": "histograms",
        "description": "Display the distribution of a data set, useful for returns or frequency distributions.",
        "styling": {
            "theme": "light",
            "bin_count": 50,
            "axes": {"stroke_width": 0.2, "grid_stroke_width": 1},
            "title": {"bold_html": True, "include": True},
            "annotations": ["x values"],
            "bar_style": {"opacity": 0.8, "corner_radius": 2},
            "number_format": {
                "apply_k_m": True,
                "thresholds": {"K": 1000, "M": 100000},
                "percentage_decimals": 2,
                "percentage_sign": True
            },
            "svg_defaults": {"height": 600, "width": 800}
        }
    },
    {
        "category": "pie_charts",
        "description": "Show composition or parts of a whole with slices representing categories.",
        "styling": {
            "theme": "light",
            "top_categories_to_show": 10,
            "bundle_rest_as": "Others",
            "title": {"bold_html": True, "include": True},
            "pie_style": {
                "inner_radius_ratio": 0.0,  # 0 for pie, >0 for donut
                "pad_angle": 0.02
            },
            "annotations": ["percentage labels"],
            "color_scheme": "d3.schemeCategory10",
            "number_format": {
                "apply_k_m": True,
                "thresholds": {"K": 1000, "M": 100000},
                "percentage_decimals": 2,
                "percentage_sign": True
            },
            "svg_defaults": {"height": 600, "width": 800}
        }
    },
    {
        "category": "tabular_and_generic_charts",
        "description": "Applies to charts where number formatting needs flexibility, including mixed or raw data.",
        "styling": {
            "theme": "light",
            "axes": {"stroke_width": 0.2, "grid_stroke_width": 1},
            "title": {"bold_html": True, "include": True},
            "annotations": ["x values"],
            "number_format": {
                "apply_k_m": True,
                "thresholds": {"K": 1000, "M": 100000},
                "exclude_if_commas_present": True,
                "exclude_if_not_numeric": True,
                "percentage_decimals": 2,
                "percentage_sign": True
            },
            "svg_defaults": {"height": 600, "width": 800}
        }
    },
    {
        "category": "heat_maps",
        "description": "Show data density or intensity using color scales on a matrix or grid.",
        "styling": {
            "theme": "light",
            "axes": {
                "stroke_color": "black",
                "stroke_width": 0.2,
                "grid_stroke_width": 1
            },
            "color_scale": "d3.interpolateViridis",
            "title": {"bold_html": True, "include": True},
            "svg_defaults": {"height": 600, "width": 800}
        }
    },
    {
        "category": "histogram_distribution",
        "description": "Specialized histogram for return distributions with opacity control.",
        "styling": {
            "theme": "light",
            "bar_style": {"opacity": 0.75},
            "axes": {
                "grid_stroke_width": 1
            },
            "title": {"bold_html": True, "include": True},
            "svg_defaults": {"height": 600, "width": 800}
        }
    }
]

# styling_instructions = [str(chart_dict).replace('\n','') for chart_dict in styling_instructions]


class GenerateVisualizationPlan(dspy.Signature):
    """
    Generate a visualization or analytical plan from a natural language query.
    The plan should include the reasoning steps and the names of charts
    (chosen from known chart types).
    """

    query = dspy.InputField(
        desc="User's natural language query or instruction, e.g., 'Compare monthly revenue trends across regions.'"
    )
    dataset_context = dspy.InputField(desc="Dataset information, columns, types etc")
    plan = dspy.OutputField(
        desc=(
            "Step-by-step plan describing what to compute or visualize, "
            "including which chart types to use. "
            "Chart names must be chosen from: "
            "[line_charts, bar_charts, histograms, pie_charts, "
            "tabular_and_generic_charts, heat_maps, histogram_distribution]."
        )
    )

    relevant_query = dspy.OutputField(
        desc="Boolean flag — True if the query is not about visualizations,dashboarding, design, data analysis etc.",
        type=bool
    )


fail = """// D3.js: No visualization generated.
// Reason: The query was not about analysis/visualization or the data is invalid.
d3.select("body")
  .append("div")
  .style("color", "red")
  .style("font-size", "1.2em")
  .style("margin", "2em")
  .text("No visualization generated: The query was not about analysis/visualization or the data is invalid.");"""

class d3_module(dspy.Module):
    def __init__(self, styling_instructions, fail):
        self.styling_instructions = styling_instructions
        self.planner = dspy.Predict(GenerateVisualizationPlan)
        self.plan_to_d3_sign = dspy.Predict(plan_to_d3)
        self.fail = fail
    def forward(self,query,dataset_context):
        plan = self.planner(query=query, dataset_context=dataset_context)

        # print(plan)
        if 'False' in str(plan.relevant_query):
            
            styling = []
            for c in self.styling_instructions:
                if c.get("category") in plan.plan:
                    styling.append(c)
            d3_code = self.plan_to_d3_sign(plan=plan.plan, dataset_context = dataset_context, styling_instructions = str(styling) )

            return d3_code.d3_js_code
        else:
            return self.fail
```

```python
import pandas as pd
import numpy as np

# Create a more realistic fake housing dataset DataFrame
np.random.seed(42)
n_samples = 20

# Bedrooms: Most homes have 2-4 bedrooms, with fewer at the extremes
bedrooms = np.random.choice([1, 2, 3, 4, 5], size=n_samples, p=[0.05, 0.25, 0.45, 0.2, 0.05])

# Bathrooms: Correlated with bedrooms, but with some noise
bathrooms = np.clip(
    np.random.normal(loc=bedrooms - 0.5 + np.random.uniform(0, 1, n_samples), scale=0.3), 1, 4
)
bathrooms = np.round(bathrooms, 1)

# Living area: log-normal for positive skew, correlated with bedrooms
sqft_living = np.random.lognormal(mean=7.5 + 0.15 * bedrooms, sigma=0.2, size=n_samples).astype(int)

# Lot size: log-normal, larger than living area, more variance
sqft_lot = np.random.lognormal(mean=8.5 + 0.1 * bedrooms, sigma=0.35, size=n_samples).astype(int)

# Floors: Most homes have 1 or 2 floors, rarely 3
floors = np.random.choice([1.0, 1.5, 2.0, 2.5, 3.0], size=n_samples, p=[0.4, 0.15, 0.35, 0.07, 0.03])

# Waterfront: Only a small fraction are waterfront
waterfront = np.random.choice([True, False], size=n_samples, p=[0.1, 0.9])

# View: Most homes have 0-1, few have 2-4
view = np.random.choice([0, 1, 2, 3, 4], size=n_samples, p=[0.5, 0.3, 0.1, 0.07, 0.03])

# Condition: Most homes are in good condition (3-4)
condition = np.random.choice([1, 2, 3, 4, 5], size=n_samples, p=[0.05, 0.15, 0.5, 0.25, 0.05])

# Grade: Most homes are 6-10, few at extremes
grade = np.random.choice(range(1, 14), size=n_samples, p=[0.01, 0.01, 0.03, 0.05, 0.08, 0.12, 0.18, 0.18, 0.15, 0.10, 0.05, 0.02, 0.02])

# Year built: Most homes built after 1970, some older
yr_built = np.random.choice(
    np.concatenate([np.arange(1950, 1980), np.arange(1980, 2000), np.arange(2000, 2023)]),
    size=n_samples,
    p=np.concatenate([
        np.repeat(0.01, 30),   # 1950-1979
        np.repeat(0.025, 20),  # 1980-1999
        np.repeat(0.025, 23)   # 2000-2022
    ]) / (0.01*30 + 0.025*43)
)

# Zipcode: Weighted to favor some zipcodes
zipcodes = ['98101', '98102', '98103', '98104', '98105']
zipcode = np.random.choice(zipcodes, size=n_samples, p=[0.2, 0.15, 0.35, 0.15, 0.15])

# Latitude and longitude: Centered around Seattle, with small noise
lat = np.round(np.random.normal(loc=47.62, scale=0.04, size=n_samples), 6)
long = np.round(np.random.normal(loc=-122.32, scale=0.04, size=n_samples), 6)

# Price: Correlated with living area, grade, and waterfront, plus noise
base_price = 100000 + sqft_living * 350 + sqft_lot * 2
grade_adj = (grade - 7) * 20000
waterfront_adj = np.where(waterfront, 150000, 0)
view_adj = view * 10000
price_noise = np.random.normal(0, 40000, n_samples)
price = base_price + grade_adj + waterfront_adj + view_adj + price_noise
price = np.round(price, -3)

housing_df = pd.DataFrame({
    "price": price,
    "bedrooms": bedrooms,
    "bathrooms": bathrooms,
    "sqft_living": sqft_living,
    "sqft_lot": sqft_lot,
    "floors": floors,
    "waterfront": waterfront,
    "view": view,
    "condition": condition,
    "grade": grade,
    "yr_built": yr_built,
    "zipcode": zipcode,
    "lat": lat,
    "long": long
})

housing_df.to_csv("housing_sample.csv")


housing_dataset_context = """
Dataset Description:
This dataset contains information about residential houses, including their sale prices and various features that may influence price. The data is commonly used for housing market analysis and price prediction.

Columns:
- price (float): Sale price of the house in USD.
- bedrooms (int): Number of bedrooms in the house.
- bathrooms (float): Number of bathrooms in the house.
- sqft_living (int): Living area in square feet.
- sqft_lot (int): Lot size in square feet.
- floors (float): Number of floors in the house.
- waterfront (bool): Whether the house has a waterfront view (True/False).
- view (int): Quality of the view from the house, rated from 0 (none) to 4 (excellent).
- condition (int): Condition of the house, rated from 1 (poor) to 5 (excellent).
- grade (int): Overall grade of the house, from 1 (low) to 13 (high).
- yr_built (int): Year the house was built.
- zipcode (str): Postal code where the house is located.
- lat (float): Latitude coordinate of the house.
- long (float): Longitude coordinate of the house.
"""

```
